### Guide to code step 2

**Load the source table.** A DataFrame is a spreadsheet-like table in memory. The path must point to the documented input file; loading a file does not validate its contents.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

DATA = Path("../data/processed")
monthly = pd.read_csv(DATA / "monthly_scorecard.csv")
products = pd.read_csv(DATA / "product_scorecard.csv")
companies = pd.read_csv(DATA / "company_scorecard.csv")
```

### Guide to code step 3

**Read the chart.** The axis labels identify units and the grouping defines what is compared. A visual association does not establish cause. Rate charts need denominators; truncated axes can exaggerate small differences.

```python
monthly["received_month"] = pd.to_datetime(monthly["received_month"])
monthly = monthly.sort_values("received_month")
figure, axis = plt.subplots(figsize=(11, 5))
axis.plot(monthly["received_month"], monthly["complaints"], color="#2563EB")
axis.set(title="Monthly Complaint Volume", ylabel="Complaints")
axis.grid(alpha=0.25)
figure.tight_layout()
```

### Guide to code step 4

**Summarize observations.** Grouping collects rows with the same label. Counts and sums produce totals; a mean of a binary 0/1 indicator gives a rate. Always identify the denominator and keep missing categories visible.

**Read the chart.** The axis labels identify units and the grouping defines what is compared. A visual association does not establish cause. Rate charts need denominators; truncated axes can exaggerate small differences.

```python
top = products.groupby("Product", as_index=False)["complaints"].sum().nlargest(12, "complaints").sort_values("complaints")
figure, axis = plt.subplots(figsize=(10, 7))
axis.barh(top["Product"], top["complaints"], color="#7C3AED")
axis.set(title="Top Products by Complaint Volume", xlabel="Complaints")
figure.tight_layout()
```

### Guide to code step 5

**Read the chart.** The axis labels identify units and the grouping defines what is compared. A visual association does not establish cause. Rate charts need denominators; truncated axes can exaggerate small differences.

```python
eligible = companies[companies["complaints"] >= 100].nsmallest(15, "timely_response_rate").sort_values("timely_response_rate")
figure, axis = plt.subplots(figsize=(10, 7))
axis.barh(eligible["Company"], eligible["timely_response_rate"], color="#DC2626")
axis.set(title="Lowest Timely Response Rates (100+ complaints)", xlabel="Rate", xlim=(0, 1))
figure.tight_layout()
```
